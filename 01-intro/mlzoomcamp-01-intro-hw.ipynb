```python
# Q5
max_ppg = df.loc[
    df.origin == "Asia",
    "fuel_efficiency_mpg"
].max()
max_ppg
```

```python
# Q6
hp_series = df["horsepower"]

hp_median_before = hp_series.median()
hp_mode = hp_series.mode().iloc[0]
hp_series = hp_series.fillna(hp_mode)
hp_median_after = hp_series.median()

print(f"median before: {hp_median_before}")
print(f"mode to fill na: {hp_mode}")
print(f"median after: {hp_median_after}")
```

```python
# Q7
cols = ["vehicle_weight", "model_year"]
X = df.loc[df.origin == "Asia", cols].head(7).to_numpy()
XTX = X.T @ X
y = [1100, 1300, 800, 900, 1000, 1100, 1200]

w = (np.linalg.inv(XTX) @ X.T) @ y
element_sum = float(w.sum())
element_sum
```
